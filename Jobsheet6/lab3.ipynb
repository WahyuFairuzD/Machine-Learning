```python
# ==============================================================================
# LAB: Support Vector Machine (SVM) - Soft Margin Tuning (Overlapping Data)
# ==============================================================================

# --- Step 1: Import Libraries & Define Plotting Function ---
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
from sklearn.svm import SVC
from sklearn.datasets import make_blobs

# Set visual style
sns.set_theme()

def plot_svc_decision_function(model, ax=None, plot_support=True):
    """Fungsi pembantu untuk memplot garis keputusan, margin, dan support vectors SVM"""
    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    # Buat grid untuk evaluasi model
    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X = np.meshgrid(y, x)
    xy = np.vstack([X.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X.shape)
    
    # Plot decision boundary (0) dan margin (-1, 1)
    ax.contour(X, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])
    
    # Plot support vectors
    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, facecolors='none', edgecolors='k')
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)


# --- Step 2: Create Dummy Data with Overlap ---
X, y = make_blobs(n_samples=100, centers=2, random_state=0, cluster_std=1.2)

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
plt.title("Step 2: Overlapping Dummy Data (cluster_std = 1.2)")
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()


# --- Step 3: Analyze the Impact of Hyperparameter Tuning (C Parameter) ---
X, y = make_blobs(n_samples=100, centers=2, random_state=0, cluster_std=0.8)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)

for axi, C in zip(ax, [10.0, 0.1]):
    model = SVC(kernel='linear', C=C).fit(X, y)
    axi.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
    plot_svc_decision_function(model, axi)
    axi.set_title('C = {0:.1f}'.format(C), size=14)

plt.show()
```
