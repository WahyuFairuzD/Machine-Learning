```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load Data
df = pd.read_csv('voice.csv')

# Preprocessing: Separation of features (X) and target (y)
X = df.drop(columns=['label'])
y = df['label']

# Encode target label ('male'/'female') to numerical values
le = LabelEncoder()
y = le.fit_transform(y)

# Feature Scaling (Crucial for SVM performance)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Define configurations
split_ratios = [0.3, 0.2]  # 70:30 and 80:20 test sizes
kernels = ['linear', 'poly', 'rbf']
results = []

# 2. Train and Evaluate Models
for test_size in split_ratios:
    train_ratio_str = f"{int((1-test_size)*100)}:{int(test_size*100)}"
    
    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled, y, test_size=test_size, random_state=42, stratify=y
    )
    
    for kernel in kernels:
        model = SVC(kernel=kernel)
        model.fit(X_train, y_train)
        
        y_pred = model.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        
        results.append({
            'Data Split (Train:Test)': train_ratio_str,
            'Kernel': kernel.upper(),
            'Accuracy': f"{acc:.4f}"
        })

# 3. Tabulate Results
results_df = pd.DataFrame(results)
print("=== SVM Performance Comparison on voice.csv ===")
print(results_df.to_string(index=False))
```
